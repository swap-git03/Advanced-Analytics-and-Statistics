```python
# claimed pattern for the qualification
# Btech  = 0.4, Mtech = 0.1, MCA = 45, MSC = 0.2
# BTech = 50, Mtech = 20, MCA = 45, MSC = 25
# whether counts obtain from sample follows the pattern or not
```

```python
counts = np.array([50,20,45,25])
pattern = np.array([0.4,0.1,0.3,0.2])
```

```python
exp_counts = sum(counts) * pattern
exp_counts
```

```python
from scipy.stats import chisquare
```

```python
scipy.stats.chisquare(counts, f_exp = exp_counts) #not rejected
```

```python
counts = np.array([70,10,35,45])
exp_counts = sum(counts) * pattern
exp_counts
```

```python
scipy.stats.chisquare(counts, f_exp = exp_counts)
# p-value<0.05, so we reject Ho
```
